```python
import numpy as np

a = np.array([[[1,2,3,4],
              [6,7,8,2],
              [9,0,1,2]],
             [[9,4,6,2],
              [1,1,1,1],
              [1,1,1,1],
             ]])
print(a.shape)
print(a.ndim)
print(a.size)
print(a.dtype)
```

```python
import numpy as np

a = np.full((2,3,4), 9)
print(a)

a = np.zeros((2,3,1))
print(a)

a = np.ones((2,3,4))
print(a)
```
