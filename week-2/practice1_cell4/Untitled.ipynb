```python
tuple = ("Ekiti", 750, 'Oshogbo', 250, "Akure", 500) 
s_tuple = ("Abeokuta", 300, "Ogbomoso") 

print(tuple) # Prints complete tuple [cite: 8]
print(tuple[-1]) # Prints last element [cite: 8]
print(tuple[2:4]) # Prints 2nd to 3rd elements [cite: 8]
print(tuple[3:]) # Prints from 3rd element onwards [cite: 9]
print(s_tuple * 3) # Prints tuple three times [cite: 9]
print(tuple + s_tuple) # Concatenates tuples [cite: 9]
```
