```python
import cmath # Using cmath to handle complex roots

print("Select the operation to perform:")
print("1. Quadratic Equation (Ax^2 + Bx + C = 0)")
print("2. Cubic Equation (Ax^3 + Bx^2 + Cx + D = 0)")
print("3. Quartic Equation (Ax^4 + Bx^3 + Cx^2 + Dx + E = 0)")

choice = input("Enter choice (1/2/3): ")

if choice == '1':
    a = float(input("Enter A: "))
    b = float(input("Enter B: "))
    c = float(input("Enter C: "))
    # Discriminant
    d = (b**2) - (4*a*c)
    root1 = (-b - cmath.sqrt(d)) / (2*a)
    root2 = (-b + cmath.sqrt(d)) / (2*a)
    print(f"The roots are {root1} and {root2}")

# Note: Cubic and Quartic solutions typically use specialized libraries 
# like numpy.roots for complex general cases.
elif choice == '2' or choice == '3':
    print("Advanced equation solving selected. Implement specialized root-finding logic here.")
else:
    print("Invalid choice")
```
