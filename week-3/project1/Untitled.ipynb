```python
import tkinter as tk
from tkinter import messagebox
import pandas as pd

# Function to verify employee
def verify_employee():
    name = entry_name.get().strip()
    dept = entry_dept.get().strip()
    
    try:
        # Load the dataset
        df = pd.read_csv('GIG-logistics.csv')
        
        # Clean column names (remove leading/trailing spaces if any)
        df.columns = df.columns.str.strip()
        
        # Filter for the specific employee
        employee = df[(df['NAME'].str.upper() == name.upper()) & 
                      (df['DEPARTMENT'].str.upper() == dept.upper())]
        
        if not employee.empty:
            # Get other members of the same department
            dept_members = df[df['DEPARTMENT'].str.upper() == dept.upper()]['NAME'].tolist()
            dept_members.remove(employee.iloc[0]['NAME']) # Remove the current user from the list
            
            members_str = "\n".join(dept_members)
            messagebox.showinfo("Success", f"Welcome {name}!\n\nOther members in {dept}:\n{members_str}")
        else:
            messagebox.showerror("Not Found", f"We are sorry, {name}. You are not listed as an employee in the {dept} department.")
            
    except FileNotFoundError:
        messagebox.showerror("Error", "The database file 'GIG-logistics.csv' was not found.")
    except Exception as e:
        messagebox.showerror("Error", f"An error occurred: {e}")

# GUI Setup
root = tk.Tk()
root.title("GIG Logistics - Name Verification")
root.geometry("400x250")

# Labels and Entry fields
tk.Label(root, text="Employee Name:", font=("Arial", 10)).pack(pady=5)
entry_name = tk.Entry(root, width=30)
entry_name.pack(pady=5)

tk.Label(root, text="Department:", font=("Arial", 10)).pack(pady=5)
entry_dept = tk.Entry(root, width=30)
entry_dept.pack(pady=5)

# Verification Button
btn_verify = tk.Button(root, text="Verify Identity", command=verify_employee, bg="blue", fg="white")
btn_verify.pack(pady=20)

root.mainloop()
```
